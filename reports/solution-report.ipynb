### 2) What difficulties did you encounter and how did you manage to solve them?

- **Difficulty:** Missing values after merging service tables (internet/phone-related columns).  
  **Solution:** Used domain-aware fills (`"No internet service"`, `"No phone service"`) instead of dropping rows.
  
- **Difficulty:** `TotalCharges` had non-numeric/blank values.  
  **Solution:** Converted to numeric with coercion and handled missing values via imputation in the preprocessing pipeline.

- **Difficulty:** Class imbalance in churn target.  
  **Solution:** Used stratified split and `class_weight='balanced'` in Logistic Regression.

- **Difficulty:** Risk of misleading metric interpretation due to imbalance.  
  **Solution:** Prioritized **AUC-ROC** for model selection and reported Accuracy as a secondary metric.
